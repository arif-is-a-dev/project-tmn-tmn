# **B. DATA PROFILING**

**1. Profiling Struktur Data**

```python
profil_struktur = []

for nama_sheet, df in data.items():

    profil_struktur.append({
        "dataset": nama_sheet,
        "jumlah_baris": df.shape[0],
        "jumlah_kolom": df.shape[1],
        "nama_kolom": ", ".join(df.columns.astype(str))
    })
profil_struktur = pd.DataFrame(profil_struktur)

# print("\n" + "="*100)
# print("B.1 PROFILING STRUKTUR DATA")
# print("="*100)

display(profil_struktur)
```

**2. Profiling Tipe Data**


```python
profil_tipe = []

for nama_sheet, df in data.items():

    for kolom in df.columns:

        profil_tipe.append({
            "dataset": nama_sheet,
            "kolom": kolom,
            "tipe_data": str(df[kolom].dtype),
            "jumlah_unique": df[kolom].nunique(dropna=True),
            "jumlah_missing": df[kolom].isna().sum()
        })

profil_tipe = pd.DataFrame(profil_tipe)

profil_tipe.head(50)
# display(profil_tipe)
```

**3. Profiling Missing Value**

```python
profil_missing = []

for nama_sheet, df in data.items():

    for kolom in df.columns:

        jumlah_missing = df[kolom].isna().sum()

        profil_missing.append({
            "dataset": nama_sheet,
            "kolom": kolom,
            "jumlah_missing": jumlah_missing,
            "persentase_missing": round(
                (jumlah_missing / len(df)) * 100, 2
            )
        })
profil_missing = pd.DataFrame(profil_missing)
profil_missing.head(50)

#display(profil_missing)
```

**Hanya tampilkan kolom yang memiliki missing value**

```python
missing_bermasalah = profil_missing[
    profil_missing["jumlah_missing"] > 0
].copy()

print("\nKolom yang memiliki missing value:")

display(missing_bermasalah)
```

**4. Profiling Duplicate**

```python
profil_duplicate = []

for nama_sheet, df in data.items():

    jumlah_duplicate = df.duplicated().sum()

    profil_duplicate.append({
        "dataset": nama_sheet,
        "jumlah_baris": len(df),
        "jumlah_duplicate": jumlah_duplicate,
        "persentase_duplicate": round(
            (jumlah_duplicate / len(df)) * 100, 2
        )
    })

profil_duplicate = pd.DataFrame(profil_duplicate)
profil_duplicate.head(50)


# display(profil_duplicate)
```

**5. Profiling Unique Value**

```python
profil_unique = []

for nama_sheet, df in data.items():

    for kolom in df.columns:

        profil_unique.append({
            "dataset": nama_sheet,
            "kolom": kolom,
            "jumlah_unique": df[kolom].nunique(dropna=True)
        })

profil_unique = pd.DataFrame(profil_unique)
profil_unique.head(50)

#display(profil_unique)
```

**6. Statistik Deskriptif**

```python
for nama_sheet, df in data.items():

    print("\n" + "-"*80)
    print(f"Dataset: {nama_sheet}")
    print("-"*80)

    # Ambil kolom numerik
    df_numerik = df.select_dtypes(include=np.number)

    if len(df_numerik.columns) > 0:
        display(df_numerik.describe().T)

    else:
        print("Tidak terdapat kolom numerik.")
```

**7. Validasi Kandidat Primary key**

```python
profil_pk = []

for nama_sheet, df in data.items():
    for kolom in df.columns:
        missing = df[kolom].isna().sum()
        duplicate = df[kolom].duplicated().sum()
        unique = df[kolom].nunique(dropna=True)

        profil_pk.append({
            "dataset": nama_sheet,
            "kolom": kolom,
            "jumlah_baris": len(df),
            "jumlah_unique": unique,
            "jumlah_missing": missing,
            "jumlah_duplicate": duplicate,
            "kandidat_primary_key": (
                missing == 0 and
                duplicate == 0 and
                unique == len(df)
            )
        })

profil_pk = pd.DataFrame(profil_pk)

display(profil_pk)

display(
    profil_pk[profil_pk["kandidat_primary_key"]]
)
```

**8. Validasi Foreign Key**

```python
hasil_fk = []

for nama_child, df_child in data.items():
    for kolom in df_child.columns:
        if "id" not in kolom.lower():
            continue

        for nama_parent, df_parent in data.items():
            if nama_child == nama_parent or kolom not in df_parent.columns:
                continue

            child = df_child[kolom].dropna()
            parent = set(df_parent[kolom].dropna())

            if child.empty:
                continue

            invalid = child[~child.isin(parent)]

            hasil_fk.append({
                "tabel_child": nama_child,
                "kolom_fk": kolom,
                "tabel_parent": nama_parent,
                "kolom_referensi": kolom,
                "jumlah_data_child": len(child),
                "jumlah_nilai_unik_child": child.nunique(),
                "jumlah_nilai_parent": len(parent),
                "jumlah_fk_tidak_ditemukan": len(invalid),
                "status_pemeriksaan": "SESUAI" if len(invalid) == 0 else "PERLU DICEK"
            })

hasil_fk = pd.DataFrame(hasil_fk)

if not hasil_fk.empty:
    display(hasil_fk)
else:
    print("Tidak ditemukan pasangan kolom ID yang dapat diperiksa antar dataset.")
```

**9. Validasi Nilai Numerik**

```python
profil_numerik = []

for nama_sheet, df in data.items():

    kolom_numerik = df.select_dtypes(
        include=np.number
    ).columns

    for kolom in kolom_numerik:

        profil_numerik.append({
            "dataset": nama_sheet,
            "kolom": kolom,
            "minimum": df[kolom].min(),
            "maksimum": df[kolom].max(),
            "rata_rata": df[kolom].mean(),
            "median": df[kolom].median(),
            "jumlah_nol": (df[kolom] == 0).sum(),
            "jumlah_negatif": (df[kolom] < 0).sum()
        })

profil_numerik = pd.DataFrame(profil_numerik)

display(profil_numerik)
```

**10. Nilai Negatif**

```python
hasil_negatif = []

for nama_sheet, df in data.items():
    for kolom in df.select_dtypes(include=np.number):
        jumlah = (df[kolom] < 0).sum()
        hasil_negatif.append({
            "dataset": nama_sheet,
            "kolom": kolom,
            "jumlah_negatif": jumlah,
            "ada_negatif": "YA" if jumlah > 0 else "TIDAK"
        })

hasil_negatif = pd.DataFrame(hasil_negatif)
display(hasil_negatif)
```

```python
for nama_sheet, df in data.items():
    for kolom in df.select_dtypes(include=np.number):
        data_negatif = df[df[kolom] < 0]

        if not data_negatif.empty:
            print(f"\n{nama_sheet} - {kolom}")
            display(data_negatif)
```

**11. Ringkasan Profiling All Dataset**

```python
ringkasan_profiling = []

for nama_sheet, df in data.items():
    jumlah_baris = len(df)
    jumlah_missing = df.isna().sum().sum()
    jumlah_duplicate = df.duplicated().sum()
    jumlah_negatif = (df.select_dtypes(include=np.number) < 0).sum().sum()

    ringkasan_profiling.append({
        "dataset": nama_sheet,
        "jumlah_baris": jumlah_baris,
        "jumlah_kolom": df.shape[1],
        "kolom_numerik": len(df.select_dtypes(include=np.number).columns),
        "total_missing": jumlah_missing,
        "persentase_missing": round(jumlah_missing / df.size * 100, 2),
        "total_duplicate": jumlah_duplicate,
        "persentase_duplicate": round(jumlah_duplicate / jumlah_baris * 100, 2),
        "total_nilai_negatif": jumlah_negatif
    })

ringkasan_profiling = pd.DataFrame(ringkasan_profiling)

display(ringkasan_profiling)
```

# **C. Data Cleaning**

```python
data_clean = {}

for nama_sheet, df in data.items():
    data_clean[nama_sheet] = df.copy()

print("Salinan data untuk proses cleaning berhasil dibuat.")
```

### Cleaning whitespace

```python
for nama_sheet, df in data_clean.items():

    for kolom in df.select_dtypes(include="object").columns:

        df[kolom] = df[kolom].apply(
            lambda x: x.strip() if isinstance(x, str) else x
        )

print("Whitespace pada data teks telah dibersihkan.")
```

###Cleaning ID

```python
kolom_id = {
    "Data_produksi_pupuk": [
        "produksi_id",
        "produk_id"
    ],

    "Data_Bahan_Organik": [
        "bahan_id",
        "ketersediaan_id",
        "padukuhan_id"
    ],

    "Data_Penjualan": [
        "Penjualan_id",
        "Produk_id"
    ],

    "Data Wilayah_Kalurahan": [
        "kalurahan_id"
    ],

    "Data_Wilayah_Padukuhan": [
        "padukuhan_id",
        "kalurahan_id"
    ],

    "Data_Pertanian": [
        "agri_id",
        "padukuhan_id"
    ]
}

for nama_sheet, kolom_list in kolom_id.items():
    if nama_sheet in data_clean:
        for kolom in kolom_list:
            if kolom in data_clean[nama_sheet].columns:
                data_clean[nama_sheet][kolom] = (
                    data_clean[nama_sheet][kolom]
                    .astype("string")
                    .str.strip()
                )
```

**4. Standardisasi Jenis/Kategori Produk**

```python
import pandas as pd
from pathlib import Path

folder = Path(r"G:\Shared drives\projek_vinix\Data_Standarisation")
# folder = Path(r"G:\Drive bersama\projek_vinix\Data_Standarisation")

# 1. Baca semua CSV
csv_files = {f.stem: pd.read_csv(f) for f in folder.glob("*.csv")}

print("=== STRUKTUR DATA ===")
for nama, df in sorted(csv_files.items()):
    print(f"\n{nama}: {df.shape[0]} baris, {df.shape[1]} kolom")
    print("Kolom:", df.columns.tolist())

# 2. Tetapkan data sesuai nama file
kal = csv_files["Data_Wilayah_Kalurahan"]
pad = csv_files["Data_Wilayah_Padukuhan"]
agri = csv_files["Data_Pertanian"]
bo = csv_files["Data_Bahan_Organik"]
produk = csv_files["Data_Produk_Katalog"]
produksi = csv_files["Data_produksi_pupuk"]
penjualan = csv_files["Data_Penjualan"]

# 3. Periksa primary key
primary_keys = {
    "KALURAHAN": (kal, "kalurahan_id"),
    "PADUKUHAN": (pad, "padukuhan_id"),
    "DATA_PERTANIAN": (agri, "agri_id"),
    "KETERSEDIAAN_BAHAN": (bo, "ketersediaan_id"),
    "PRODUK": (produk, "produk_id"),
    "PRODUKSI": (produksi, "produksi_id"),
    "PENJUALAN": (penjualan, "Penjualan_id"),
}

print("\n=== PEMERIKSAAN PRIMARY KEY ===")
for nama, (df, kolom) in primary_keys.items():
    ada = kolom in df.columns
    if not ada:
        print(f"{nama}: GAGAL — kolom {kolom} tidak ditemukan")
        continue

    null = int(df[kolom].isna().sum())
    duplikat = int(df[kolom].duplicated().sum())
    print(
        f"{nama}.{kolom}: "
        f"null={null}, duplikat={duplikat}, "
        f"unik={null == 0 and duplikat == 0}"
    )

# 4. Periksa foreign key terhadap tabel induk
def cek_fk(nama, child, kolom_child, parent, kolom_parent):
    if kolom_child not in child.columns or kolom_parent not in parent.columns:
        print(f"{nama}: GAGAL — kolom relasi tidak ditemukan")
        return

    child_values = child[kolom_child].dropna().astype(str).str.strip()
    parent_values = set(
        parent[kolom_parent].dropna().astype(str).str.strip()
    )
    yatim = sorted(set(child_values) - parent_values)

    print(
        f"{nama}: nilai kosong={child[kolom_child].isna().sum()}, "
        f"nilai yatim={len(yatim)}"
    )
    if yatim:
        print("  Contoh nilai yatim:", yatim[:10])

print("\n=== PEMERIKSAAN FOREIGN KEY ===")
cek_fk("PADUKUHAN → KALURAHAN",
       pad, "kalurahan_id", kal, "kalurahan_id")

cek_fk("DATA_PERTANIAN → PADUKUHAN",
       agri, "padukuhan_id", pad, "padukuhan_id")

cek_fk("KETERSEDIAAN_BAHAN → PADUKUHAN",
       bo, "padukuhan_id", pad, "padukuhan_id")

cek_fk("PRODUKSI → PRODUK",
       produksi, "produk_id", produk, "produk_id")

cek_fk("PENJUALAN → PRODUK",
       penjualan, "Produk_id", produk, "produk_id")

# 5. Periksa apakah bahan_id konsisten dengan nama dan kategori
print("\n=== PEMERIKSAAN BAHAN ORGANIK ===")
wajib = ["bahan_id", "jenis_bahan_organik", "kategori_bahan"]
if all(c in bo.columns for c in wajib):
    konsistensi = (
        bo.groupby("bahan_id")[["jenis_bahan_organik", "kategori_bahan"]]
        .nunique(dropna=False)
    )
    tidak_konsisten = konsistensi[
        (konsistensi > 1).any(axis=1)
    ]

    print("Jumlah bahan_id unik:", bo["bahan_id"].nunique())
    print("Jumlah bahan dengan nama/kategori tidak konsisten:",
          len(tidak_konsisten))
    if not tidak_konsisten.empty:
        print(tidak_konsisten)
else:
    print("GAGAL — kolom bahan organik tidak lengkap.")

print("\nValidasi selesai. Belum ada data yang ditulis ke database.")
```

**Versi 2 — ETL saat tabel sudah berisi data**

```python
# ETL Versi 2: Tambahkan data baru tanpa memasukkan duplikat
import pandas as pd
from sqlalchemy import MetaData, Table
from sqlalchemy.dialects.postgresql import insert

metadata = MetaData()
metadata.reflect(bind=engine, schema="public")

with engine.begin() as conn:
    for nama in urutan_tabel:
        df = data_db[nama]

        if df.empty:
            print(f"{nama}: tidak ada data untuk dimasukkan.")
            continue

        tabel = Table(
            nama,
            metadata,
            schema="public",
            autoload_with=conn
        )

        # Pastikan semua kolom sumber tersedia di tabel target
        kolom_tabel = {kolom.name for kolom in tabel.columns}
        kolom_df = set(df.columns)

        if not kolom_df.issubset(kolom_tabel):
            raise ValueError(
                f"Kolom pada data {nama} tidak sesuai dengan tabel PostgreSQL."
            )

        # Ubah NaN/NaT menjadi None agar dikenali sebagai NULL oleh PostgreSQL
        df_simpan = df.astype(object).where(pd.notna(df), None)
        records = df_simpan.to_dict(orient="records")

        # Masukkan baris baru; abaikan baris yang melanggar
        # primary key atau unique constraint yang sudah ada
        stmt = insert(tabel).on_conflict_do_nothing()

        for awal in range(0, len(records), 100):
            batch = records[awal:awal + 100]
            conn.execute(stmt, batch)

        print(f"{nama}: proses penambahan data selesai.")

print("ETL versi 2 selesai.")
```

## Validasi langsung di PostgreSQL

```python
from sqlalchemy import text

# 1. Periksa jumlah baris di semua tabel
print("=== A. RECORD COUNT ===")

with engine.connect() as conn:
    for nama, df in data_db.items():
        aktual = conn.execute(
            text(f"SELECT COUNT(*) FROM public.{nama}")
        ).scalar_one()

        print(
            f"{nama}: DB={aktual}, "
            f"DataFrame={len(df)}, "
            f"OK={aktual == len(df)}"
        )

# 2. Periksa NULL dan duplikasi primary key di database
print("\n=== B. NULL DAN DUPLIKASI PRIMARY KEY ===")

pk_map = {
    "kalurahan": "kalurahan_id",
    "padukuhan": "padukuhan_id",
    "data_pertanian": "agri_id",
    "bahan_organik": "bahan_id",
    "ketersediaan_bahan": "ketersediaan_id",
    "produk": "produk_id",
    "produksi": "produksi_id",
    "penjualan": "penjualan_id",
}

with engine.connect() as conn:
    for tabel, pk in pk_map.items():
        hasil = conn.execute(text(f"""
            SELECT
                COUNT(*) FILTER (WHERE {pk} IS NULL) AS null_id,
                COUNT(*) - COUNT(DISTINCT {pk}) AS duplikasi_id
            FROM public.{tabel}
        """)).one()

        print(
            f"{tabel}: NULL PK={hasil[0]}, "
            f"duplikasi PK={hasil[1]}"
        )

# 3. Periksa constraint PK dan FK yang terpasang di PostgreSQL
print("\n=== C. CONSTRAINT DATABASE ===")

with engine.connect() as conn:
    constraints = conn.execute(text("""
        SELECT
            tc.table_name,
            tc.constraint_type,
            tc.constraint_name
        FROM information_schema.table_constraints tc
        WHERE tc.table_schema = 'public'
          AND tc.table_name IN (
              'kalurahan', 'padukuhan', 'data_pertanian',
              'bahan_organik', 'ketersediaan_bahan',
              'produk', 'produksi', 'penjualan'
          )
          AND tc.constraint_type IN ('PRIMARY KEY', 'FOREIGN KEY')
        ORDER BY tc.table_name, tc.constraint_type
    """)).fetchall()

    for row in constraints:
        print(f"{row[0]} | {row[1]} | {row[2]}")
```

## Uji foreign key, JOIN, dan agregasi

```python
from sqlalchemy import text

# D. Periksa nilai yatim pada seluruh relasi
print("=== D. FOREIGN KEY ORPHAN CHECK ===")

queries_fk = {
    "padukuhan -> kalurahan": """
        SELECT COUNT(*)
        FROM padukuhan c
        LEFT JOIN kalurahan p ON c.kalurahan_id = p.kalurahan_id
        WHERE p.kalurahan_id IS NULL
    """,
    "data_pertanian -> padukuhan": """
        SELECT COUNT(*)
        FROM data_pertanian c
        LEFT JOIN padukuhan p ON c.padukuhan_id = p.padukuhan_id
        WHERE p.padukuhan_id IS NULL
    """,
    "ketersediaan_bahan -> padukuhan": """
        SELECT COUNT(*)
        FROM ketersediaan_bahan c
        LEFT JOIN padukuhan p ON c.padukuhan_id = p.padukuhan_id
        WHERE p.padukuhan_id IS NULL
    """,
    "ketersediaan_bahan -> bahan_organik": """
        SELECT COUNT(*)
        FROM ketersediaan_bahan c
        LEFT JOIN bahan_organik p ON c.bahan_id = p.bahan_id
        WHERE p.bahan_id IS NULL
    """,
    "produksi -> produk": """
        SELECT COUNT(*)
        FROM produksi c
        LEFT JOIN produk p ON c.produk_id = p.produk_id
        WHERE p.produk_id IS NULL
    """,
    "penjualan -> produk": """
        SELECT COUNT(*)
        FROM penjualan c
        LEFT JOIN produk p ON c.produk_id = p.produk_id
        WHERE p.produk_id IS NULL
    """
}

with engine.connect() as conn:
    for nama, query in queries_fk.items():
        jumlah = conn.execute(text(query)).scalar_one()
        print(f"{nama}: orphan={jumlah} -> {'OK' if jumlah == 0 else 'PERLU DIPERIKSA'}")


# E. Uji JOIN dan jumlah record yang terhubung
print("\n=== E. JOIN CHECK ===")

queries_join = {
    "Kalurahan dengan padukuhan": """
        SELECT COUNT(*)
        FROM padukuhan p
        JOIN kalurahan k ON p.kalurahan_id = k.kalurahan_id
    """,
    "Pertanian dengan padukuhan": """
        SELECT COUNT(*)
        FROM data_pertanian a
        JOIN padukuhan p ON a.padukuhan_id = p.padukuhan_id
    """,
    "Ketersediaan dengan bahan dan padukuhan": """
        SELECT COUNT(*)
        FROM ketersediaan_bahan kb
        JOIN bahan_organik b ON kb.bahan_id = b.bahan_id
        JOIN padukuhan p ON kb.padukuhan_id = p.padukuhan_id
    """,
    "Produksi dengan produk": """
        SELECT COUNT(*)
        FROM produksi pr
        JOIN produk p ON pr.produk_id = p.produk_id
    """,
    "Penjualan dengan produk": """
        SELECT COUNT(*)
        FROM penjualan pj
        JOIN produk p ON pj.produk_id = p.produk_id
    """
}

with engine.connect() as conn:
    for nama, query in queries_join.items():
        jumlah = conn.execute(text(query)).scalar_one()
        print(f"{nama}: {jumlah} record terhubung")


# F. Agregasi penjualan per produk
print("\n=== F. AGREGASI PENJUALAN PER PRODUK ===")

with engine.connect() as conn:
    hasil = conn.execute(text("""
        SELECT
            p.produk_id,
            p.nama_produk,
            COUNT(pj.penjualan_id) AS jumlah_record_penjualan,
            COALESCE(SUM(pj.jumlah_terjual), 0) AS total_terjual,
            COALESCE(SUM(pj.total_penjualan), 0) AS total_nilai_penjualan
        FROM produk p
        LEFT JOIN penjualan pj ON p.produk_id = pj.produk_id
        GROUP BY p.produk_id, p.nama_produk
        ORDER BY p.produk_id
    """)).fetchall()

    for row in hasil:
        print(row)

print("\nPemeriksaan selesai; tidak ada data yang diubah.")
```

Tahap 4 — Menghitung bobot AHP

Kegunaan: AHP menentukan tingkat kepentingan relatif dari empat kriteria. Bobotnya nanti dipakai oleh SAW.

```python

# STAGE 40 — Pembobotan kriteria dengan AHP

import numpy as np
import pandas as pd

kode_kriteria = ["K1", "K2", "K3", "K4"]

nama_kriteria = {
    "K1": "Ketersediaan bahan",
    "K2": "Kontinuitas",
    "K3": "Jarak ke lokasi produksi",
    "K4": "Kemudahan pengumpulan",
}

# Matriks perbandingan berpasangan AHP.
# CATATAN: nilai berikut merupakan ASUMSI SIMULASI,
# bukan hasil wawancara mitra atau penilaian ahli.
matriks_ahp = np.array([
    [1,   2,   2,   2],
    [1/2, 1,   1,   2],
    [1/2, 1,   1,   1],
    [1/2, 1/2, 1,   1],
], dtype=float)

# Validasi matriks
assert matriks_ahp.shape == (4, 4)
assert np.all(matriks_ahp > 0)
assert np.allclose(np.diag(matriks_ahp), 1)
assert np.allclose(matriks_ahp, matriks_ahp.T ** -1)

# Menghitung nilai eigen dan vektor eigen
nilai_eigen, vektor_eigen = np.linalg.eig(matriks_ahp)

indeks_maks = np.argmax(nilai_eigen.real)
lambda_maks = float(nilai_eigen[indeks_maks].real)

vektor_utama = np.abs(vektor_eigen[:, indeks_maks].real)
bobot_ahp = vektor_utama / vektor_utama.sum()

# Tabel hasil bobot
tabel_bobot_ahp = pd.DataFrame({
    "kode_kriteria": kode_kriteria,
    "nama_kriteria": [
        nama_kriteria[k] for k in kode_kriteria
    ],
    "bobot": bobot_ahp,
})

print("=== MATRIKS PERBANDINGAN AHP ===")
display(pd.DataFrame(
    matriks_ahp,
    index=kode_kriteria,
    columns=kode_kriteria,
))

print("\n=== HASIL BOBOT AHP ===")
display(tabel_bobot_ahp.round(5))

print(f"\nTotal bobot: {bobot_ahp.sum():.5f}")
print(f"Lambda maksimum: {lambda_maks:.5f}")
```

Tahap 4.1 — Menguji konsistensi AHP


Kegunaan: memeriksa apakah penilaian berpasangan dalam matriks AHP cukup konsisten secara matematis.

```python

# STAGE 41 — Uji konsistensi AHP

n = len(kode_kriteria)

CI = (lambda_maks - n) / (n - 1)

# Random Index (RI) untuk ukuran matriks 1 sampai 10
nilai_RI = {
    1: 0.00,
    2: 0.00,
    3: 0.58,
    4: 0.90,
    5: 1.12,
    6: 1.24,
    7: 1.32,
    8: 1.41,
    9: 1.45,
    10: 1.49,
}

RI = nilai_RI[n]
CR = CI / RI if RI > 0 else 0.0

print("Lambda maksimum:", round(lambda_maks, 5))
print("Consistency Index (CI):", round(CI, 5))
print("Random Index (RI):", round(RI, 5))
print("Consistency Ratio (CR):", round(CR, 5))

if CR <= 0.10:
    print("Konsistensi matriks memenuhi patokan CR <= 0.10.")
else:
    print(
        "Konsistensi belum memenuhi patokan. "
        "Tinjau kembali dasar perbandingan kriteria."
    )
```

Patokan CR ≤ 0,10 lazim digunakan untuk menilai konsistensi matriks AHP. Namun, matriks yang konsisten secara matematis belum tentu mencerminkan preferensi mitra yang sebenarnya. Kita tetap harus menyebut sumber bobotnya secara jujur.

**Tahap 5 — Menghitung ranking SAW menggunakan bobot AHP**

Kegunaan: menggabungkan empat kriteria menjadi satu skor akhir per bahan. Berbeda dari pengujian sebelumnya, tahap ini memakai bobot hasil perhitungan AHP, bukan bobot sama rata 25%.

Rumus normalisasi SAW yang digunakan:
- Benefit: nilai alternatif dibagi nilai maksimum kriteria.
- Cost: nilai minimum kriteria dibagi nilai alternatif.
- Skor akhir: jumlah nilai normalisasi yang sudah dikalikan bobot masing-masing.

```python

# STAGE 42 — Perhitungan SAW dengan bobot AHP

# Pastikan urutan bobot sesuai urutan kriteria
hasil_saw_final = ringkasan_saw.copy()
bobot_final = dict(zip(kode_kriteria, bobot_ahp))

hasil_saw_final = ringkasan_saw.copy()

# Normalisasi benefit
hasil_saw_final["norm_ketersediaan"] = (
    hasil_saw_final["jumlah_ketersediaan"]
    / hasil_saw_final["jumlah_ketersediaan"].max()
)

hasil_saw_final["norm_kontinuitas"] = (
    hasil_saw_final["skor_kontinuitas"]
    / hasil_saw_final["skor_kontinuitas"].max()
)

hasil_saw_final["norm_kemudahan"] = (
    hasil_saw_final["skor_kemudahan"]
    / hasil_saw_final["skor_kemudahan"].max()
)

# Normalisasi cost
hasil_saw_final["norm_jarak"] = (
    hasil_saw_final["jarak_ke_produksi_km"].min()
    / hasil_saw_final["jarak_ke_produksi_km"]
)

# Hitung skor akhir dengan bobot AHP
hasil_saw_final["skor_saw"] = (
    bobot_final["K1"]
    * hasil_saw_final["norm_ketersediaan"]
    + bobot_final["K2"]
    * hasil_saw_final["norm_kontinuitas"]
    + bobot_final["K3"]
    * hasil_saw_final["norm_jarak"]
    + bobot_final["K4"]
    * hasil_saw_final["norm_kemudahan"]
)

# Urutkan skor tertinggi ke terendah
hasil_saw_final = hasil_saw_final.sort_values(
    "skor_saw",
    ascending=False,
).reset_index(drop=True)

hasil_saw_final.insert(
    0,
    "peringkat",
    range(1, len(hasil_saw_final) + 1),
)

kolom_hasil = [
    "peringkat",
    "bahan_id",
    "jenis_bahan_organik",
    "kategori_bahan",
    "skor_saw",
    "jumlah_ketersediaan",
    "jarak_ke_produksi_km",
    "skor_kontinuitas",
    "skor_kemudahan",
]

display(hasil_saw_final[kolom_hasil].round(2))
```

**Tahap 7 — Menguji kestabilan ranking**

Kegunaan: melihat apakah peringkat bahan berubah jika bobot kriteria berubah. Ini penting karena bobot AHP yang kita gunakan sekarang masih asumsi peneliti.

Untuk uji awal, kita bandingkan ranking dengan bobot AHP dan bobot sama rata. Bobot sama rata di sini hanya menjadi skenario pembanding.

```python

# STAGE 44 — Perbandingan ranking SAW berbobot sama rata
# dengan ranking SAW berbobot AHP

import numpy as np
import pandas as pd

# 1. Pastikan variabel sumber tersedia

assert "ringkasan_saw" in globals(), (
    "ringkasan_saw tidak tersedia. Jalankan Stage 39."
)

assert "hasil_saw_final" in globals(), (
    "hasil_saw_final tidak tersedia. Jalankan Stage 42."
)

# 2. Buat ulang hasil SAW berbobot sama rata

hasil_saw = ringkasan_saw.copy()

# Bobot sama rata untuk empat kriteria
bobot_sama = {
    "K1": 0.25,  # Ketersediaan
    "K2": 0.25,  # Kontinuitas
    "K3": 0.25,  # Jarak
    "K4": 0.25,  # Kemudahan pengumpulan
}

# Normalisasi kriteria benefit
hasil_saw["norm_ketersediaan"] = (
    hasil_saw["jumlah_ketersediaan"]
    / hasil_saw["jumlah_ketersediaan"].max()
)

hasil_saw["norm_kontinuitas"] = (
    hasil_saw["skor_kontinuitas"]
    / hasil_saw["skor_kontinuitas"].max()
)

hasil_saw["norm_kemudahan"] = (
    hasil_saw["skor_kemudahan"]
    / hasil_saw["skor_kemudahan"].max()
)

# Normalisasi kriteria cost
hasil_saw["norm_jarak"] = (
    hasil_saw["jarak_ke_produksi_km"].min()
    / hasil_saw["jarak_ke_produksi_km"]
)

# Hitung skor SAW berbobot sama rata
hasil_saw["skor_saw_uji"] = (
    bobot_sama["K1"] * hasil_saw["norm_ketersediaan"]
    + bobot_sama["K2"] * hasil_saw["norm_kontinuitas"]
    + bobot_sama["K3"] * hasil_saw["norm_jarak"]
    + bobot_sama["K4"] * hasil_saw["norm_kemudahan"]
)

# Urutkan dari skor tertinggi ke terendah
hasil_saw = hasil_saw.sort_values(
    "skor_saw_uji",
    ascending=False
).reset_index(drop=True)

# Buat peringkat pengujian
hasil_saw["peringkat_uji"] = (
    hasil_saw["skor_saw_uji"]
    .rank(method="min", ascending=False)
    .astype(int)
)

# 3. Ambil peringkat SAW berbobot sama rata

ranking_uji = hasil_saw[
    ["bahan_id", "peringkat_uji"]
].copy()

ranking_uji = ranking_uji.rename(
    columns={
        "peringkat_uji": "peringkat_bobot_sama"
    }
)

# 4. Ambil peringkat SAW berbobot AHP

ranking_ahp = hasil_saw_final[
    ["bahan_id", "peringkat"]
].copy()

ranking_ahp = ranking_ahp.rename(
    columns={
        "peringkat": "peringkat_bobot_ahp"
    }
)

# 5. Gabungkan dan bandingkan ranking

perbandingan_ranking = ranking_uji.merge(
    ranking_ahp,
    on="bahan_id",
    how="inner",
    validate="one_to_one"
)

# Positif = peringkat turun
# Negatif = peringkat naik
# Nol = peringkat tetap
perbandingan_ranking["perubahan_posisi"] = (
    perbandingan_ranking["peringkat_bobot_ahp"]
    - perbandingan_ranking["peringkat_bobot_sama"]
)

perbandingan_ranking = perbandingan_ranking.sort_values(
    "peringkat_bobot_ahp"
).reset_index(drop=True)

# 6. Validasi hasil perbandingan

assert len(hasil_saw) == 17
assert len(hasil_saw_final) == 17
assert len(perbandingan_ranking) == 17

assert hasil_saw["bahan_id"].is_unique
assert hasil_saw_final["bahan_id"].is_unique
assert perbandingan_ranking["bahan_id"].is_unique

assert perbandingan_ranking[
    "peringkat_bobot_sama"
].notna().all()

assert perbandingan_ranking[
    "peringkat_bobot_ahp"
].notna().all()

print("PERBANDINGAN RANKING: BERHASIL")
print("Jumlah alternatif:", len(perbandingan_ranking))

display(perbandingan_ranking)
```

Interpretasinya:
- perubahan_posisi = 0: posisi bahan tidak berubah.
- Nilai positif: posisi bahan turun setelah memakai bobot AHP.
- Nilai negatif: posisi bahan naik setelah memakai bobot AHP.

```python
#STAGE 46.1 — VALIDASI GRANULARITAS PRODUK-BULAN

df_granular_ml = df_target_ml.copy()

# Pastikan kolom identitas observasi tersedia
kolom_granularitas = ["Produk", "Bulan"]

assert all(
    kolom in df_granular_ml.columns
    for kolom in kolom_granularitas
), "Kolom Produk atau Bulan tidak ditemukan."

# Pastikan identitas produk dan bulan tidak kosong
assert df_granular_ml[kolom_granularitas].notna().all().all(), (
    "Ada nilai Produk atau Bulan yang kosong."
)

# Hitung duplikasi kombinasi produk-bulan
jumlah_duplikat = df_granular_ml.duplicated(
    subset=["Produk", "Bulan"]
).sum()

jumlah_produk = df_granular_ml["Produk"].nunique()
jumlah_bulan = df_granular_ml["Bulan"].dt.to_period("M").nunique()
jumlah_observasi = len(df_granular_ml)

print("=== STAGE 46.1: VALIDASI GRANULARITAS ===")
print("Granularitas       : Satu produk per bulan")
print("Jumlah produk      :", jumlah_produk)
print("Jumlah bulan       :", jumlah_bulan)
print("Jumlah observasi   :", jumlah_observasi)
print("Duplikasi produk-bulan:", jumlah_duplikat)

assert jumlah_duplikat == 0, (
    "Terdapat duplikasi kombinasi produk-bulan."
)

print("\nValidasi granularitas berhasil.")
```

```python
# STAGE 46.2 — PEMERIKSAAN KELENGKAPAN URUTAN BULAN

hasil_per_produk = []

for produk, grup in df_granular_ml.groupby("Produk"):
    bulan_produk = (
        grup["Bulan"]
        .dt.to_period("M")
        .drop_duplicates()
        .sort_values()
    )

    bulan_lengkap = pd.period_range(
        start=bulan_produk.min(),
        end=bulan_produk.max(),
        freq="M"
    )

    bulan_hilang = bulan_lengkap.difference(bulan_produk)

    hasil_per_produk.append({
        "Produk": produk,
        "bulan_awal": str(bulan_produk.min()),
        "bulan_akhir": str(bulan_produk.max()),
        "jumlah_bulan_tercatat": len(bulan_produk),
        "jumlah_bulan_hilang": len(bulan_hilang),
        "daftar_bulan_hilang": ", ".join(
            str(bulan) for bulan in bulan_hilang
        )
    })

df_cek_bulan = pd.DataFrame(hasil_per_produk)

display(
    df_cek_bulan.sort_values("Produk").reset_index(drop=True)
)

print(
    "\nTotal bulan hilang:",
    df_cek_bulan["jumlah_bulan_hilang"].sum()
)
```

```python

# STAGE 48.2 — PEMBENTUKAN DATASET FITUR AWAL

df_model_ml = df_granular_ml.copy()

# Urutkan waktu di dalam setiap produk
df_model_ml = df_model_ml.sort_values(
    ["Produk", "Bulan"]
).reset_index(drop=True)

# Fitur kalender
df_model_ml["bulan_kalender"] = (
    df_model_ml["Bulan"].dt.month
)

# Fitur historis penjualan per produk
for lag in [1, 2, 3, 12]:
    df_model_ml[f"lag_{lag}"] = (
        df_model_ml.groupby("Produk")["Jumlah_Terjual"]
        .shift(lag)
    )

# Daftar fitur dan target
fitur_final_awal = [
    "Produk",
    "bulan_kalender",
    "lag_1",
    "lag_2",
    "lag_3",
    "lag_12",
]

target_ml = "Jumlah_Terjual"

print("=== FITUR AWAL MODEL ===")
print("Fitur:", fitur_final_awal)
print("Target:", target_ml)

print("\n=== CONTOH DATA ===")
display(
    df_model_ml[
        [
            "Produk",
            "Bulan",
            "Jumlah_Terjual",
            "bulan_kalender",
            "lag_1",
            "lag_2",
            "lag_3",
            "lag_12",
        ]
    ].head(15)
)

print("\n=== NILAI KOSONG AKIBAT PEMBENTUKAN LAG ===")
display(
    df_model_ml[fitur_final_awal]
    .isna()
    .sum()
    .to_frame("jumlah_kosong")
)
```

Stage 53 — Latih dan evaluasi 5 kandidat model

```python
# STAGE 53 — EVALUASI KANDIDAT MODEL PADA VALIDASI 2024

import numpy as np
import pandas as pd
from sklearn.base import clone
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.pipeline import make_pipeline

# Data dan fitur dasar
df = df_model_siap.copy()
df["Bulan"] = pd.to_datetime(df["Bulan"])
df = df.sort_values(["Produk", "Bulan"])

fitur_dasar = [
    "Produk", "bulan_kalender", "lag_1", "lag_2", "lag_3", "lag_12"
]

# Rolling hanya memakai bulan sebelumnya
g = df.groupby("Produk")["Jumlah_Terjual"]
df["rolling_mean_3"] = g.transform(
    lambda s: s.shift(1).rolling(3, min_periods=3).mean()
)
df["rolling_mean_12"] = g.transform(
    lambda s: s.shift(1).rolling(12, min_periods=12).mean()
)
fitur_rolling = fitur_dasar + ["rolling_mean_3", "rolling_mean_12"]

train = df[df["Bulan"] < "2024-01-01"].dropna(
    subset=fitur_rolling + ["Jumlah_Terjual"]
)
valid = df[
    (df["Bulan"] >= "2024-01-01") &
    (df["Bulan"] < "2025-01-01")
].dropna(subset=fitur_rolling + ["Jumlah_Terjual"])

def buat_pipeline(model, fitur):
    prep = ColumnTransformer([
        ("produk", OneHotEncoder(handle_unknown="ignore"), ["Produk"]),
        ("angka", "passthrough", [f for f in fitur if f != "Produk"])
    ])
    return make_pipeline(prep, model)

model_specs = [
    ("Random Forest", RandomForestRegressor(
        n_estimators=300, min_samples_leaf=2, random_state=42
    ), fitur_dasar),
    ("RF + Rolling 3 & 12", RandomForestRegressor(
        n_estimators=300, min_samples_leaf=2, random_state=42
    ), fitur_rolling),
    ("Extra Trees", ExtraTreesRegressor(
        n_estimators=300, min_samples_leaf=2, random_state=42
    ), fitur_dasar),
    ("HistGradientBoosting", HistGradientBoostingRegressor(
        max_iter=200, random_state=42
    ), fitur_dasar)
]

try:
    from xgboost import XGBRegressor
    model_specs.append(("XGBoost", XGBRegressor(
        n_estimators=300, max_depth=4, learning_rate=0.05,
        objective="reg:squarederror", random_state=42
    ), fitur_dasar))
except ImportError:
    print("XGBoost dilewati: library belum terpasang.")

hasil = []
model_terlatih = {}

for nama, estimator, fitur in model_specs:
    # RF + rolling butuh 12 bulan histori; evaluasi validasi tetap sama
    tr = train.dropna(subset=fitur + ["Jumlah_Terjual"])
    va = valid.dropna(subset=fitur + ["Jumlah_Terjual"])

    pipe = buat_pipeline(clone(estimator), fitur)
    pipe.fit(tr[fitur], tr["Jumlah_Terjual"])
    pred = pipe.predict(va[fitur])
    aktual = va["Jumlah_Terjual"]

    hasil.append({
        "Model": nama,
        "Jumlah_validasi": len(va),
        "MAE": mean_absolute_error(aktual, pred),
        "RMSE": np.sqrt(mean_squared_error(aktual, pred)),
        "WAPE (%)": np.abs(aktual - pred).sum() / np.abs(aktual).sum() * 100
    })
    model_terlatih[nama] = pipe

hasil_model_validasi = pd.DataFrame(hasil).round(3)
display(hasil_model_validasi.sort_values("WAPE (%)").reset_index(drop=True))
```

Interpretasi utamanya:
- Extra Trees unggul pada MAE dan WAPE. Artinya, dari model yang diuji, kesalahan absolut rata-rata dan kesalahan relatif terhadap total penjualan paling rendah.
- XGBoost memiliki RMSE paling rendah, yaitu 7,581, sehingga sedikit lebih baik dalam metrik yang memberi penalti lebih besar pada kesalahan ekstrem.
- RF + Rolling 3 & 12 tidak unggul pada validasi ini. Hasilnya lebih buruk daripada RF biasa berdasarkan MAE dan WAPE. Ini menunjukkan bahwa fitur rolling tidak otomatis meningkatkan akurasi.
- Baseline Lag-12 sebelumnya memiliki WAPE 64,036%, sehingga semua lima model pada tabel ini mengungguli baseline tersebut dalam WAPE pada validasi 2024.

Stage 59 — Prediksi penjualan 12 bulan ke depan (Januari–Desember 2026)

```python

# Stage 59 — Prediksi penjualan 12 bulan ke depan

import pandas as pd
import numpy as np

fitur_final = [
    "Produk",
    "bulan_kalender",
    "lag_1",
    "lag_2",
    "lag_3",
    "lag_12"
]

df_hist = df_model_siap.copy()
df_hist["Bulan"] = pd.to_datetime(df_hist["Bulan"])

produk_list = sorted(df_hist["Produk"].unique())
tanggal_terakhir = df_hist["Bulan"].max()

hasil_prediksi = []

for produk in produk_list:
    histori = (
        df_hist[df_hist["Produk"] == produk]
        .sort_values("Bulan")
        .set_index("Bulan")["Jumlah_Terjual"]
        .astype(float)
        .to_dict()
    )

    for i in range(1, 13):
        tanggal = tanggal_terakhir + pd.DateOffset(months=i)

        lag_1 = histori.get(tanggal - pd.DateOffset(months=1), np.nan)
        lag_2 = histori.get(tanggal - pd.DateOffset(months=2), np.nan)
        lag_3 = histori.get(tanggal - pd.DateOffset(months=3), np.nan)
        lag_12 = histori.get(tanggal - pd.DateOffset(months=12), np.nan)

        baris = pd.DataFrame([{
            "Produk": produk,
            "bulan_kalender": tanggal.month,
            "lag_1": lag_1,
            "lag_2": lag_2,
            "lag_3": lag_3,
            "lag_12": lag_12
        }])[fitur_final]

        prediksi = float(model_final.predict(baris)[0])
        prediksi = max(0, prediksi)

        hasil_prediksi.append({
            "Bulan": tanggal,
            "Produk": produk,
            "Prediksi_Penjualan": prediksi
        })

        # Simpan prediksi agar menjadi lag untuk bulan selanjutnya
        histori[tanggal] = prediksi

df_prediksi_2026 = pd.DataFrame(hasil_prediksi)

df_prediksi_2026["Prediksi_Penjualan"] = (
    df_prediksi_2026["Prediksi_Penjualan"].round(2)
)

display(df_prediksi_2026.head(15))
print("Jumlah prediksi:", len(df_prediksi_2026))
print("Jumlah produk:", df_prediksi_2026["Produk"].nunique())
print(
    "Periode:",
    df_prediksi_2026["Bulan"].min().date(),
    "sampai",
    df_prediksi_2026["Bulan"].max().date()
)
```

```python
# STAGE 58 — Analisis Kesalahan Prediksi Extra Trees pada Data Uji 2025

import numpy as np
import pandas as pd

# Membuat tabel analisis prediksi
analisis_error_2025 = uji_2025[
    ["Bulan", "Produk", "Jumlah_Terjual"]
].copy()

analisis_error_2025["Prediksi"] = prediksi_2025

# Menghitung kesalahan prediksi
analisis_error_2025["Error"] = (
    analisis_error_2025["Jumlah_Terjual"]
    - analisis_error_2025["Prediksi"]
)

analisis_error_2025["Absolute_Error"] = (
    analisis_error_2025["Error"].abs()
)

# Persentase kesalahan absolut
analisis_error_2025["APE (%)"] = np.where(
    analisis_error_2025["Jumlah_Terjual"] != 0,
    analisis_error_2025["Absolute_Error"]
    / analisis_error_2025["Jumlah_Terjual"].abs()
    * 100,
    np.nan
)

# Mengurutkan berdasarkan kesalahan terbesar
analisis_error_2025 = analisis_error_2025.sort_values(
    "Absolute_Error",
    ascending=False
).reset_index(drop=True)

print("=== 10 PREDIKSI DENGAN KESALAHAN TERBESAR ===")

display(
    analisis_error_2025.head(10).round(3)
)

print("\n=== RINGKASAN KESALAHAN PER PRODUK ===")

ringkasan_error_produk = (
    analisis_error_2025
    .groupby("Produk")
    .agg(
        Jumlah_Data=("Jumlah_Terjual", "count"),
        Total_Aktual=("Jumlah_Terjual", "sum"),
        Total_Prediksi=("Prediksi", "sum"),
        MAE=("Absolute_Error", "mean"),
        Total_Error=("Error", "sum")
    )
    .reset_index()
)

ringkasan_error_produk = ringkasan_error_produk.sort_values(
    "MAE",
    ascending=False
).reset_index(drop=True)

display(ringkasan_error_produk.round(3))

print("\n=== RINGKASAN ARAH KESALAHAN ===")

jumlah_prediksi_terlalu_rendah = (
    analisis_error_2025["Error"] > 0
).sum()

jumlah_prediksi_terlalu_tinggi = (
    analisis_error_2025["Error"] < 0
).sum()

jumlah_prediksi_sama = (
    analisis_error_2025["Error"] == 0
).sum()

print("Prediksi terlalu rendah:", jumlah_prediksi_terlalu_rendah)
print("Prediksi terlalu tinggi:", jumlah_prediksi_terlalu_tinggi)
print("Prediksi sama dengan aktual:", jumlah_prediksi_sama)
```

# Hyperparameter Tuning Extra Trees

```python
# STAGE 60 — Hyperparameter Tuning Extra Trees
# Pemilihan konfigurasi berdasarkan validasi tahun 2024

import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.pipeline import Pipeline

# --------------------------------------------------
# 1. Siapkan data train dan validasi
# --------------------------------------------------

fitur = [
    "Produk",
    "bulan_kalender",
    "lag_1",
    "lag_2",
    "lag_3",
    "lag_12"
]

target = "Jumlah_Terjual"

train_tuning = train.dropna(
    subset=fitur + [target]
).copy()

valid_tuning = valid.dropna(
    subset=fitur + [target]
).copy()

X_train = train_tuning[fitur]
y_train = train_tuning[target]

X_valid = valid_tuning[fitur]
y_valid = valid_tuning[target]

print("Jumlah data train:", len(X_train))
print("Jumlah data validasi:", len(X_valid))

# --------------------------------------------------
# 2. Ambil preprocessing dari model Extra Trees lama
# --------------------------------------------------

model_et_lama = model_terlatih["Extra Trees"]

# Pertahankan preprocessing yang sebelumnya digunakan.
# Ambil estimator preprocessing dan buat kandidat baru.
preprocessor = clone(model_et_lama.steps[0][1])

# --------------------------------------------------
# 3. Tentukan konfigurasi yang akan diuji
# --------------------------------------------------

konfigurasi = [
    {
        "n_estimators": 200,
        "max_depth": None,
        "min_samples_leaf": 1,
        "max_features": 1.0
    },
    {
        "n_estimators": 300,
        "max_depth": None,
        "min_samples_leaf": 2,
        "max_features": 1.0
    },
    {
        "n_estimators": 500,
        "max_depth": None,
        "min_samples_leaf": 2,
        "max_features": 1.0
    },
    {
        "n_estimators": 300,
        "max_depth": 10,
        "min_samples_leaf": 2,
        "max_features": 1.0
    },
    {
        "n_estimators": 300,
        "max_depth": 15,
        "min_samples_leaf": 3,
        "max_features": 1.0
    },
    {
        "n_estimators": 300,
        "max_depth": None,
        "min_samples_leaf": 3,
        "max_features": 0.8
    },
    {
        "n_estimators": 500,
        "max_depth": 15,
        "min_samples_leaf": 2,
        "max_features": 0.8
    }
]

# --------------------------------------------------
# 4. Latih dan evaluasi setiap konfigurasi
# --------------------------------------------------

hasil_tuning = []
model_tuning = {}

for nomor, parameter in enumerate(konfigurasi, start=1):

    preprocessor_kandidat = clone(preprocessor)

    estimator = ExtraTreesRegressor(
        **parameter,
        random_state=42,
        n_jobs=-1
    )

    pipeline = Pipeline([
        ("preprocessor", preprocessor_kandidat),
        ("model", estimator)
    ])

    pipeline.fit(X_train, y_train)

    prediksi_valid = pipeline.predict(X_valid)

    mae = mean_absolute_error(
        y_valid,
        prediksi_valid
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_valid,
            prediksi_valid
        )
    )

    penyebut_wape = np.abs(y_valid).sum()

    wape = (
        np.abs(y_valid.to_numpy() - prediksi_valid).sum()
        / penyebut_wape
        * 100
        if penyebut_wape != 0
        else np.nan
    )

    nama_model = f"ET_Tuning_{nomor}"

    model_tuning[nama_model] = pipeline

    hasil_tuning.append({
        "Model": nama_model,
        **parameter,
        "Jumlah_validasi": len(y_valid),
        "MAE": mae,
        "RMSE": rmse,
        "WAPE (%)": wape
    })

hasil_tuning = pd.DataFrame(hasil_tuning)

hasil_tuning = hasil_tuning.sort_values(
    "WAPE (%)",
    ascending=True
).reset_index(drop=True)

print("\n=== HASIL HYPERPARAMETER TUNING ===")

display(
    hasil_tuning.style.format({
        "MAE": "{:.3f}",
        "RMSE": "{:.3f}",
        "WAPE (%)": "{:.3f}"
    })
)

# --------------------------------------------------
# 5. Tampilkan konfigurasi terbaik berdasarkan WAPE
# --------------------------------------------------

baris_terbaik = hasil_tuning.iloc[0]

nama_model_terbaik_tuning = baris_terbaik["Model"]

print("\n=== KONFIGURASI TERBAIK PADA VALIDASI 2024 ===")
print("Nama model:", nama_model_terbaik_tuning)
print("MAE:", round(baris_terbaik["MAE"], 3))
print("RMSE:", round(baris_terbaik["RMSE"], 3))
print("WAPE (%):", round(baris_terbaik["WAPE (%)"], 3))

print("\nParameter terbaik:")
print({
    k: baris_terbaik[k]
    for k in [
        "n_estimators",
        "max_depth",
        "min_samples_leaf",
        "max_features"
    ]
})
```

# Prediksi Penjualan Januari–Desember 2026

```python
# STAGE 63 — Prediksi Penjualan Januari–Desember 2026
# Model: Extra Trees hasil hyperparameter tuning

import pandas as pd
import numpy as np

# --------------------------------------------------
# 1. Menentukan fitur model
# --------------------------------------------------

fitur_final = [
    "Produk",
    "bulan_kalender",
    "lag_1",
    "lag_2",
    "lag_3",
    "lag_12"
]

# --------------------------------------------------
# 2. Menyiapkan data historis
# --------------------------------------------------

df_hist = df_model_siap.copy()

df_hist["Bulan"] = pd.to_datetime(df_hist["Bulan"])

df_hist = df_hist.sort_values(
    ["Produk", "Bulan"]
).copy()

produk_list = sorted(df_hist["Produk"].unique())

tanggal_terakhir = df_hist["Bulan"].max()

print("Tanggal historis terakhir:", tanggal_terakhir.date())
print("Jumlah produk:", len(produk_list))

# --------------------------------------------------
# 3. Membuat prediksi secara berurutan
# --------------------------------------------------

hasil_prediksi = []

for produk in produk_list:

    histori = (
        df_hist[df_hist["Produk"] == produk]
        .sort_values("Bulan")
        .set_index("Bulan")["Jumlah_Terjual"]
        .astype(float)
        .to_dict()
    )

    for i in range(1, 13):

        tanggal = tanggal_terakhir + pd.DateOffset(
            months=i
        )

        # Membentuk fitur lag berdasarkan riwayat yang tersedia
        lag_1 = histori.get(
            tanggal - pd.DateOffset(months=1),
            np.nan
        )

        lag_2 = histori.get(
            tanggal - pd.DateOffset(months=2),
            np.nan
        )

        lag_3 = histori.get(
            tanggal - pd.DateOffset(months=3),
            np.nan
        )

        lag_12 = histori.get(
            tanggal - pd.DateOffset(months=12),
            np.nan
        )

        # Memastikan seluruh fitur tersedia
        nilai_lag = [lag_1, lag_2, lag_3, lag_12]

        if not np.isfinite(nilai_lag).all():
            raise ValueError(
                f"Fitur lag tidak lengkap untuk produk "
                f"{produk} pada {tanggal:%Y-%m}"
            )

        # Membuat satu baris untuk diprediksi
        baris_prediksi = pd.DataFrame([{
            "Produk": produk,
            "bulan_kalender": tanggal.month,
            "lag_1": lag_1,
            "lag_2": lag_2,
            "lag_3": lag_3,
            "lag_12": lag_12
        }])[fitur_final]

        # Prediksi menggunakan model final hasil tuning
        prediksi = float(
            model_final.predict(baris_prediksi)[0]
        )

        # Penjualan tidak boleh negatif
        prediksi = max(0.0, prediksi)

        hasil_prediksi.append({
            "Bulan": tanggal,
            "Produk": produk,
            "Prediksi_Penjualan": prediksi
        })

        # Simpan prediksi agar menjadi riwayat
        # untuk menghitung lag bulan berikutnya
        histori[tanggal] = prediksi

# --------------------------------------------------
# 4. Membuat tabel prediksi
# --------------------------------------------------

df_prediksi_2026 = pd.DataFrame(hasil_prediksi)

df_prediksi_2026["Prediksi_Penjualan"] = (
    df_prediksi_2026["Prediksi_Penjualan"].round(2)
)

df_prediksi_2026 = df_prediksi_2026.sort_values(
    ["Bulan", "Produk"]
).reset_index(drop=True)

# --------------------------------------------------
# 5. Validasi hasil
# --------------------------------------------------

print("\n=== HASIL PREDIKSI TAHUN 2026 ===")

display(df_prediksi_2026.head(15))

print("\nJumlah baris prediksi:", len(df_prediksi_2026))
print("Jumlah produk:", df_prediksi_2026["Produk"].nunique())
print(
    "Periode prediksi:",
    df_prediksi_2026["Bulan"].min().strftime("%Y-%m"),
    "sampai",
    df_prediksi_2026["Bulan"].max().strftime("%Y-%m")
)

print(
    "Total prediksi penjualan 2026:",
    round(df_prediksi_2026["Prediksi_Penjualan"].sum(), 2)
)

# Pemeriksaan kelengkapan
assert len(df_prediksi_2026) == 132, (
    "Jumlah prediksi seharusnya 132 baris."
)

assert df_prediksi_2026["Produk"].nunique() == 11, (
    "Jumlah produk seharusnya 11."
)

assert df_prediksi_2026["Bulan"].nunique() == 12, (
    "Jumlah bulan prediksi seharusnya 12."
)

assert df_prediksi_2026["Prediksi_Penjualan"].notna().all(), (
    "Masih ada prediksi yang kosong."
)

assert (
    df_prediksi_2026["Prediksi_Penjualan"] >= 0
).all(), "Ditemukan prediksi negatif."

print("\nVALIDASI BERHASIL: 132 prediksi tersedia.")
```
